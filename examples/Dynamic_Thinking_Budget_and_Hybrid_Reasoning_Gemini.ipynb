### 📖 Overview
Google's frontier Gemini reasoning models introduce **Hybrid Thinking Capabilities**, allowing developers to control the depth of internal chain-of-thought exploration before output generation.

This recipe demonstrates how to:
1. Configure **Thinking Budget parameters** (`thinking_config`) for fine-grained latency/accuracy trade-offs.
2. Solve complex multi-step **Cyber Security algorithmic verification, cryptographic puzzles, and distributed systems logic**.
3. Inspect and stream explicit **thought traces** vs final outputs.
4. Benchmark performance and token efficiency across shallow vs deep thinking budgets.

### Initialize Client

```python
import os
from google import genai
from google.genai import types

api_key = os.environ.get("GEMINI_API_KEY", "YOUR_API_KEY_HERE")
client = genai.Client(api_key=api_key)
```

## ⚙️ 1. Configuring Dynamic Thinking Budgets

We configure `thinking_config` inside `GenerateContentConfig` to allocate thinking tokens.

```python
# Complex Security & Distributed Systems Challenge Problem
complex_challenge_prompt = (
    "Analyze the following Zero-Knowledge Proof (ZKP) and Byzantine Fault Tolerant (BFT) consensus problem: \n"
    "In a network of 3f + 1 validator nodes where up to f nodes can be malicious or collude, "
    "how do we mathematically guarantee liveness and non-repudiation while verifying batch Merkle proofs? "
    "Provide a formal mathematical proof, step-by-step security reasoning, and optimal timeout bounds."
)

# 1. High-Reasoning Deep Thought Call
deep_thinking_config = types.GenerateContentConfig(
    thinking_config=types.ThinkingConfig(
        thinking_budget_token_count=2048  # Allocate deep reasoning budget
    ),
    temperature=0.2,
)

print("[*] Sending request to Gemini Reasoning Engine...")
response = client.models.generate_content(
    model="gemini-2.0-flash-thinking-exp",
    contents=complex_challenge_prompt,
    config=deep_thinking_config,
)

print("=" * 60)
print("🧠 REASONING OUTPUT:")
print("=" * 60)
print(response.text[:1000] + "...")
```

## 📊 2. Streaming Thoughts in Real Time

With streaming, you can observe the model's intermediate thoughts as they are synthesized.

```python
coding_challenge = """
Implement an ultra-optimized LRU (Least Recently Used) cache with O(1) time complexity 
for both get() and put() operations in Python, supporting thread-safe atomic lock primitives. 
Explain edge cases for concurrent race conditions.
"""

print("[*] Streaming reasoning tokens:\n")
stream_response = client.models.generate_content_stream(
    model="gemini-2.0-flash-thinking-exp",
    contents=coding_challenge,
    config=deep_thinking_config
)

for chunk in stream_response:
    if chunk.text:
        print(chunk.text, end="", flush=True)
```
